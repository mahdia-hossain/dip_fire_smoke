# Fire / smoke detection demo (firesmoke package)
Adaptive DIP preprocessing + MobileViT-XS. Upload a photo and see every DIP step and the answer.

**Runtime:** CPU is fine; GPU is faster.

```python
REPO = "mahdia-hossain/dip_fire_smoke"   # change if the repo is renamed
WEIGHTS = "firesmoke_CA.pt"
TAG = "v1.0"
!pip install -q git+https://github.com/{REPO}
!wget -q -O {WEIGHTS} https://github.com/{REPO}/releases/download/{TAG}/{WEIGHTS}
!ls -lh {WEIGHTS}
```

```python
from google.colab import files
uploaded = files.upload()
photo = list(uploaded)[0]
```

```python
from firesmoke import FireSmokeDetector
model = FireSmokeDetector(WEIGHTS)
result = model.predict(photo)
result.explain()
result.show_steps()
```

```python
out = result.save("out")
print(sorted(p.name for p in out.iterdir()))
```
